Total Exp Graph

```python
import plotly.graph_objects as go

# Define 4 integers
integers = [400634772, 352743748, 73080582, 248797500]
labels = ['Helltides', 'University', 'Pits 115', 'Infernal Hordes']

# Define custom colors for each bar
colors = ['#1f77b4', 'red', '#2ca02c', 'Yellow'] # Example colors

# Create a bar trace with custom colors
fig = go.Figure(
    data=[go.Bar(x=labels, y=integers, marker_color=colors)],
    layout=go.Layout(
        title=go.layout.Title(text="Diablo 4: Total Exp Farming Per Hour")
    )
)

# Display the plot
fig.show()
```
