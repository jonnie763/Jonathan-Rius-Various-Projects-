```python
import matplotlib.pyplot as plt
import pandas as pd

# Assuming 'df' is available from previous cells. If not, you might need to redefine it.
# For example:
data = {'Category': ['HellTides', 'University', 'Pits 115', 'Infernal Hordes'],
        'Value': [400634772, 352743748, 73080582, 248797500]}
df = pd.DataFrame(data)

# Divide the 'Value' column by 60 to get values per minute
df['Value'] = df['Value'] / 60

# Define a list of custom colors for each bar
custom_colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'] # Example custom colors

plt.figure(figsize=(8, 5))
bars = plt.bar(df['Category'], df['Value'], color=custom_colors)

# Add annotations on top of the bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.5, f'{yval:.0f}', ha='center', va='bottom')

plt.title("Diablo 4: Total Exp Farming Per Minute")
plt.xlabel('Category')
plt.ylabel('Value (Per Minute)')
plt.grid(axis='y', linestyle='--')
plt.show()
```
